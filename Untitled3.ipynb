```python
# 1. Cargar el archivo Excel desde Colab
file_path = 'AutoKnowledge_100_Reportes.xlsx'
df = pd.read_excel(file_path, skiprows=5, header=None)
df.columns = ['ID', 'Reporte']
df = df[df['ID'] != 'ID'].dropna(subset=['ID']).reset_index(drop=True)

# 2. Seleccionar 15 reportes de muestra (5 de cada área aproximadamente según tus datos)
# Vamos a tomar los primeros 15 para la demostración y tabla comparativa
muestra_df = df.head(15).copy()

# 3. Función de preprocesamiento de texto
def preprocesar_texto(texto):
    doc = nlp(texto.lower())
    tokens = [token.text for token in doc if token.is_alpha] # Tokenización básica y limpieza de puntuación

    # Palabras vacías en español + conservación de negaciones clave si es necesario
    stop_words = set(stopwords.words('spanish'))

    # Lematización y eliminación de stopwords (conservando negaciones importantes)
    tokens_procesados = []
    for token in doc:
        if token.is_alpha:
            palabra_lema = token.lemma_.lower()
            # Filtramos stopwords pero cuidamos términos clave o negaciones
            if palabra_lema not in stop_words or palabra_lema in ['no', 'sin', 'nunca']:
                tokens_procesados.append(palabra_lema)

    return tokens, tokens_procesados

# Aplicar a la muestra
muestra_df['Tokens'], muestra_df['Texto_Preprocesado'] = zip(*muestra_df['Reporte'].apply(preprocesar_texto))

# Mostrar tabla comparativa de resultados
print("--- TABLA COMPARATIVA DE PREPROCESAMIENTO (Muestra de 15 reportes) ---")
tabla_resultado = muestra_df[['ID', 'Reporte', 'Texto_Preprocesado']].copy()
tabla_resultado['Texto_Preprocesado'] = tabla_resultado['Texto_Preprocesado'].apply(lambda x: ' '.join(x))
display(tabla_resultado)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
import nltk
from nltk.corpus import stopwords

# Asegurar que tenemos las stopwords de nltk descargadas
nltk.download('stopwords')
spanish_stopwords = list(stopwords.words('spanish'))

# 1. Cargar el archivo y preparar los datos
file_path = 'AutoKnowledge_100_Reportes.xlsx'
df = pd.read_excel(file_path, skiprows=5, header=None)
df.columns = ['ID', 'Reporte']
df = df[df['ID'] != 'ID'].dropna(subset=['ID']).reset_index(drop=True)

# Asignar categorías balanceadas para los 100 reportes
def asignar_categoria_completa(index):
    if index < 34:
        return "Finanzas"
    elif index < 67:
        return "Soporte Técnico"
    else:
        return "Recursos Humanos"

df['Categoría'] = [asignar_categoria_completa(i) for i in range(len(df))]

# 2. División de datos: 80% entrenamiento, 20% pruebas (estratificado)
X = df['Reporte']
y = df['Categoría']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 3. Transformación TF-IDF con la lista de stopwords de nltk
vectorizer = TfidfVectorizer(max_features=1000, stop_words=spanish_stopwords)
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

# 4. Mostrar resultados en pantalla
print("--- DIVISIÓN Y VECTORIZACIÓN COMPLETADA EXITOSAMENTE ---")
print(f"Total de documentos de entrenamiento: {X_train.shape[0]} (80%)")
print(f"Total de documentos de prueba: {X_test.shape[0]} (20%)")
print(f"Dimensiones de la matriz TF-IDF de entrenamiento: {X_train_tfidf.shape}")
print(f"Dimensiones de la matriz TF-IDF de pruebas: {X_test_tfidf.shape}")

print("\n--- DISTRIBUCIÓN EN CONJUNTO DE ENTRENAMIENTO ---")
display(y_train.value_counts())

print("\n--- DISTRIBUCIÓN EN CONJUNTO DE PRUEBAS ---")
display(y_test.value_counts())
```

```python
from sklearn.svm import SVC
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, accuracy_score

# 1. Inicializar y entrenar Support Vector Machines (SVM)
svm_model = SVC(kernel='linear', random_state=42)
svm_model.fit(X_train_tfidf, y_train)

# 2. Inicializar y entrenar Naive Bayes
nb_model = MultinomialNB()
nb_model.fit(X_train_tfidf, y_train)

# 3. Realizar predicciones con el conjunto de pruebas
y_pred_svm = svm_model.predict(X_test_tfidf)
y_pred_nb = nb_model.predict(X_test_tfidf)

# 4. Calcular la precisión (Accuracy) de ambos modelos
acc_svm = accuracy_score(y_test, y_pred_svm)
acc_nb = accuracy_score(y_test, y_pred_nb)

print("--- RESULTADOS COMPARATIVOS DE ENTRENAMIENTO ---")
print(f"Precisión (Accuracy) de SVM: {acc_svm * 100:.2f}%")
print(f"Precisión (Accuracy) de Naive Bayes: {acc_nb * 100:.2f}%")

print("\n--- REPORTE DETALLADO DE CLASIFICACIÓN: SVM ---")
print(classification_report(y_test, y_pred_svm))

print("\n--- REPORTE DETALLADO DE CLASIFICACIÓN: NAIVE BAYES ---")
print(classification_report(y_test, y_pred_nb))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# Generar la matriz de confusión para SVM
cm = confusion_matrix(y_test, y_pred_svm, labels=svm_model.classes_)

# Graficar la matriz de confusión
plt.figure(figsize=(7, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=svm_model.classes_,
            yticklabels=svm_model.classes_)
plt.title('Matriz de Confusión - Modelo SVM (AutoKnowledge)', fontsize=12, fontweight='bold')
plt.xlabel('Predicción del Modelo', fontsize=10)
plt.ylabel('Valor Real (Ground Truth)', fontsize=10)
plt.show()
```

```python
# Instalar sentence-transformers si no está instalado
!pip install -q sentence-transformers

import pandas as pd
from sentence_transformers import SentenceTransformer

# 1. Cargar el archivo con los 100 reportes (si ya tienes df cargado, puedes omitir esta recarga)
file_path = 'AutoKnowledge_100_Reportes.xlsx'
df = pd.read_excel(file_path, skiprows=5, header=None)
df.columns = ['ID', 'Reporte']
df = df[df['ID'] != 'ID'].dropna(subset=['ID']).reset_index(drop=True)

# 2. Seleccionar un modelo multilingüe preentrenado compatible con español
# Usaremos 'paraphrase-multilingual-MiniLM-L12-v2' optimizado para similitud semántica
model = SentenceTransformer('paraphrase-multilingual-MiniLM-L12-v2')

# 3. Generar los embeddings para los 100 reportes corporativos
textos_reportes = df['Reporte'].tolist()
embeddings = model.encode(textos_reportes)

# 4. Comprobar que cada documento tenga una representación numérica
print("--- COMPROBACIÓN DE EMBEDDINGS ---")
print(f"Total de reportes procesados: {len(textos_reportes)}")
print(f"Total de matrices vectoriales generadas: {len(embeddings)}")
print(f"Dimensiones del arreglo de embeddings: {embeddings.shape}")

# Mostrar el vector del primer reporte como evidencia numérica
print(f"\nDimensiones del vector del primer reporte (ID: {df['ID'][0]}): {embeddings[0].shape}")
print(f"Primeros 5 valores del vector embedding:\n{embeddings[0][:5]}")
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

def buscar_reportes_semanticos(consulta_usuario, top_n=3):
    # 1. Transformar la consulta en un embedding usando el mismo modelo
    vector_consulta = model.encode([consulta_usuario])

    # 2. Comparar el vector de la consulta con los 100 vectores almacenados (embeddings)
    similitudes = cosine_similarity(vector_consulta, embeddings)[0]

    # 3. Obtener los índices de los documentos ordenados por mayor similitud
    indices_ordenados = np.argsort(similitudes)[::-1]

    # 4. Construir un DataFrame con los resultados más relevantes
    resultados = []
    for idx in indices_ordenados[:top_n]:
        resultados.append({
            "ID": df['ID'][idx],
            "Reporte": df['Reporte'][idx],
            "Similitud (Coseno)": round(float(similitudes[idx]), 4)
        })

    df_resultados = pd.DataFrame(resultados)
    return df_resultados

# Probar la función con el ejemplo solicitado
consulta_ejemplo = "Un empleado no puede acceder a su correo empresarial"
print(f"--- CONSULTA DEL USUARIO: '{consulta_ejemplo}' ---\n")
df_resultado_busqueda = buscar_reportes_semanticos(consulta_ejemplo, top_n=3)
display(df_resultado_busqueda)
```
