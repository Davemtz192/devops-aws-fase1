```python
import pandas as pd
import numpy as np

# 2. Cargar el archivo
df = pd.read_csv('dataset_videojuegos_1000.csv')

# 3. Muestras y estructuras básicas
print("--- PRIMERAS 5 FILAS ---")
display(df.head())

print("\n--- ÚLTIMAS 5 FILAS ---")
display(df.tail())

print("\n--- CANTIDAD DE FILAS Y COLUMNAS ---")
print(df.shape)

print("\n--- NOMBRES DE LAS VARIABLES ---")
print(df.columns.tolist())

print("\n--- TIPO DE DATO DE CADA COLUMNA ---")
print(df.dtypes)

print("\n--- ESTADÍSTICAS DESCRIPTIVAS (NUMÉRICAS) ---")
display(df.describe())

print("\n--- ESTADÍSTICAS DESCRIPTIVAS (CATEGÓRICAS) ---")
display(df.describe(include=['O']))

# 4. Identificación de problemas de calidad
print("\n--- VALORES FALTANTES POR COLUMNA ---")
print(df.isnull().sum())

print("\n--- REGISTROS DUPLICADOS ---")
print(f"Total duplicados: {df.duplicated().sum()}")

# Verificar categorías y posibles inconsistencias
print("\n--- CATEGORÍAS EN MARCA ---")
print(df['Marca'].unique())

print("\n--- CATEGORÍAS EN PLATAFORMA ---")
print(df['Plataforma'].unique())

# 5. Análisis de la distribución de Nivel_exito
print("\n--- DISTRIBUCIÓN DE NIVEL DE ÉXITO ---")
print(df['Nivel_exito'].value_counts())
print(df['Nivel_exito'].value_counts(normalize=True) * 100)
```

```python
# ==========================================
# PARTE 2: LIMPIEZA Y PREPARACIÓN DE DATOS
# ==========================================

# 1. VALORES FALTANTES: Cálculo de cantidad y porcentaje
print("--- 1. VALORES FALTANTES (CANTIDAD Y PORCENTAJE) ---")
nulos_cant = df.isnull().sum()
nulos_porc = (df.isnull().sum() / len(df)) * 100
df_nulos = pd.DataFrame({'Cantidad_Nulos': nulos_cant, 'Porcentaje_%': nulos_porc})
print(df_nulos[df_nulos['Cantidad_Nulos'] > 0])

# Imputación de valores faltantes según estrategia justificada
df['Precio_USD'].fillna(df['Precio_USD'].median(), inplace=True)
df['Horas_duracion'].fillna(df['Horas_duracion'].median(), inplace=True)
df['Tamano_GB'].fillna(df['Tamano_GB'].median(), inplace=True)
df['Marketing_millones_USD'].fillna(df['Marketing_millones_USD'].median(), inplace=True)
df['Resenas_usuarios'].fillna(df['Resenas_usuarios'].median(), inplace=True)
df['Puntuacion_critica'].fillna(df['Puntuacion_critica'].median(), inplace=True)
df['Puntuacion_usuarios'].fillna(df['Puntuacion_usuarios'].median(), inplace=True)


# 2. REGISTROS DUPLICADOS: Identificación, ejemplos y eliminación
print("\n--- 2. REGISTROS DUPLICADOS ---")
print(f"Cantidad de duplicados antes de limpiar: {df.duplicated().sum()}")

# Mostrar algunos ejemplos de duplicados si existen
if df.duplicated().sum() > 0:
    print("Ejemplos de registros duplicados:")
    display(df[df.duplicated(keep=False)].head(4))

# Filas antes de eliminar
filas_antes = len(df)

# Eliminar duplicados
df.drop_duplicates(inplace=True)

# Filas después de eliminar
filas_despues = len(df)
print(f"Filas antes de eliminar: {filas_antes}")
print(f"Filas después de eliminar: {filas_despues}")


# 3. CATEGORÍAS INCONSISTENTES: Estandarización de texto
print("\n--- 3. CATEGORÍAS INCONSISTENTES (MARCA ANTES) ---")
print(df['Marca'].unique())

# Limpieza: pasar a minúsculas, quitar espacios y estandarizar nombres comerciales
df['Marca'] = df['Marca'].astype(str).str.strip().str.title()
df['Marca'] = df['Marca'].replace({'Pc': 'PC'})

print("\n--- CATEGORÍAS EN MARCA DESPUÉS DE LA LIMPIEZA ---")
print(df['Marca'].unique())


# 4. VALORES ATÍPICOS (OUTLIERS): Análisis y decisión de mantenimiento
print("\n--- 4. VALORES ATÍPICOS ---")
print("Se evaluaron mediante el describe() previo. Se decidió *mantenerse* ya que representan valores reales de la industria de videojuegos (ej. juegos muy largos de 240 horas o presupuestos altos de marketing de 350M USD que corresponden a títulos AAA legítimos).")
```

```python
# ==========================================
# PARTE 4: PREPARACIÓN PARA EL MODELADO
# ==========================================
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# 1 y 2. Definir variables predictoras (X) y objetivo (y)
# Excluimos la variable objetivo y columnas de identificación/texto libre si las hubiera
X = df.drop(columns=['ID_Videojuego', 'Titulo', 'Nivel_exito'])
y = df['Nivel_exito']

# 3 y 4. División de datos: 70% entrenamiento, 30% prueba con estratificación y semilla
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    stratify=y,
    random_state=42
)

print(f"Dimensiones de X_train: {X_train.shape}")
print(f"Dimensiones de X_test: {X_test.shape}")

# Identificar columnas numéricas y categóricas automáticamente
columnas_numericas = X_train.select_dtypes(include=['int64', 'float64']).columns
columnas_categoricas = X_train.select_dtypes(include=['object', 'category']).columns

# 5, 6 y 7. Preparación mediante Pipeline y ColumnTransformer
transformador_numerico = Pipeline(steps=[
    ('imputador', SimpleImputer(strategy='median')),
    ('escalador', StandardScaler())
])

transformador_categorico = Pipeline(steps=[
    ('imputador', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

preprocesador = ColumnTransformer(
    transformers=[
        ('num', transformador_numerico, columnas_numericas),
        ('cat', transformador_categorico, columnas_categoricas)
    ]
)

# Aplicar el preprocesamiento ajustándose ÚNICAMENTE a los datos de entrenamiento (evita fuga de datos)
X_train_procesado = preprocesador.fit_transform(X_train)
X_test_procesado = preprocesador.transform(X_test)

print("\n¡Preprocesamiento completado exitosamente sin fuga de información!")
print(f"Forma de los datos de entrenamiento procesados: {X_train_procesado.shape}")
```

```python
# ==========================================
# PARTE 5: MODELO INICIAL (LÍNEA BASE)
# ==========================================
import time
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. Construir un Pipeline que integre el preprocesamiento y el modelo inicial
modelo_inicial = Pipeline(steps=[
    ('preprocesador', preprocesador),
    ('clasificador', RandomForestClassifier(random_state=42))
])

# 2. Medir el tiempo de entrenamiento
inicio = time.time()
modelo_inicial.fit(X_train, y_train)
fin = time.time()
tiempo_entrenamiento = fin - inicio

# 3. Generar predicciones y métricas
y_pred = modelo_inicial.predict(X_test)
exactitud = accuracy_score(y_test, y_pred)

print(f"--- REGISTRO DEL MODELO INICIAL ---")
print(f"Algoritmo utilizado: Random Forest")
print(f"Parámetros iniciales: random_state=42 (valores por defecto)")
print(f"Cantidad de variables después de la codificación: {X_train_procesado.shape[1]}")
print(f"Tiempo de entrenamiento: {tiempo_entrenamiento:.4f} segundos")
print(f"Exactitud (Accuracy): {exactitud:.4f}\n")
print("Reporte de Clasificación:")
print(classification_report(y_test, y_pred))
```

```python
# ==========================================
# PARTE 6: SELECCIÓN DE CARACTERÍSTICAS
# ==========================================
from sklearn.feature_selection import SelectFromModel
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
import pandas as pd

# 1 y 2. Identificar y seleccionar características usando la importancia del Random Forest base
selector = SelectFromModel(
    estimator=modelo_inicial.named_steps['clasificador'],
    threshold='median', # Selecciona las variables cuya importancia esté por encima de la mediana
    prefit=True
)

# Obtener nombres de las características transformadas
cat_encoder = modelo_inicial.named_steps['preprocesador'].named_transformers_['cat'].named_steps['onehot']
nombres_cat = list(cat_encoder.get_feature_names_out(columnas_categoricas))
todas_caracteristicas = list(columnas_numericas) + nombres_cat

# 3. Mostrar las variables seleccionadas
mascara = selector.get_support()
caracteristicas_seleccionadas = [col for col, keep in zip(todas_caracteristicas, mascara) if keep]

print(f"Total de variables originales: {len(todas_caracteristicas)}")
print(f"Total de variables seleccionadas: {len(caracteristicas_seleccionadas)}\n")
print("Variables seleccionadas:")
for var in caracteristicas_seleccionadas:
    print(f" - {var}")

# 5. Entrenar nuevamente el modelo con el subconjunto de características
X_train_sel = selector.transform(X_train_procesado)
X_test_sel = selector.transform(X_test_procesado)

modelo_reducido = RandomForestClassifier(random_state=42)
modelo_reducido.fit(X_train_sel, y_train)

# 6. Comparar resultados con el modelo inicial
y_pred_red = modelo_reducido.predict(X_test_sel)
exactitud_inicial = 0.9024
exactitud_reducida = accuracy_score(y_test, y_pred_red)

print(f"\n--- COMPARACIÓN DE RESULTADOS ---")
print(f"Exactitud Modelo Inicial (49 variables): {exactitud_inicial:.4f}")
print(f"Exactitud Modelo Reducido ({len(caracteristicas_seleccionadas)} variables): {exactitud_reducida:.4f}")
```
